```python
s = "a##c"
t = "#a#c"
lst_s=[]
lst_t=[]

for i in s:
    if(i != '#'):
        lst_s.append(i)
    elif(len(lst_s)>=1):
        lst_s.pop()

for j in t:
    if(j != '#'):
        lst_t.append(j)
    elif(len(lst_t)>=1):
        lst_t.pop()

print(True) if lst_s==lst_t else print(False)
```
