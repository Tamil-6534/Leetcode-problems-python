```python
nums1 = [4,9,5]
nums2 = [9,4,9,8,4]
lst=[]

if(len(nums1)<=len(nums2)):
    for i in range(len(nums1)):
        for j in range(len(nums2)):
            if(nums1[i]==nums2[j]):
                lst.append(nums1[i])
                nums2.pop(j)
                nums2.insert(j,'*')
                break

else:
    for i in range(len(nums2)):
        for j in range(len(nums1)):
            if(nums2[i]==nums1[j]):
                lst.append(nums2[i])
                nums1.pop(j)
                nums1.insert(j,'*')
                break


print(lst)
```
