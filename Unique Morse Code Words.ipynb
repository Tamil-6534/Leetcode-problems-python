```python
words = ["gin","zen","gig","msg"]

morse_dict = {
    'a': '.-',    'b': '-...',  'c': '-.-.',  'd': '-..',
    'e': '.',     'f': '..-.',  'g': '--.',   'h': '....',
    'i': '..',    'j': '.---',  'k': '-.-',   'l': '.-..',
    'm': '--',    'n': '-.',    'o': '---',   'p': '.--.',
    'q': '--.-',  'r': '.-.',   's': '...',   't': '-',
    'u': '..-',   'v': '...-',  'w': '.--',   'x': '-..-',
    'y': '-.--',  'z': '--..'
}

morse_lst=[]
for i in words:
    morse=''
    for j in i:
        morse+=morse_dict[j]
    morse_lst.append(morse)

result=set(morse_lst)
len(result)
```
