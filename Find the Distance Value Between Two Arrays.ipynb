```python
arr1 = [1,4,2,3]
arr2 = [-4,-3,6,10,20,30]
d = 3
count=0
arr1.sort()
arr2.sort()
print(arr1,arr2)
for i  in range(len(arr1)):
    ans=0
    for j in range(len(arr2)):
        
        calc=arr1[i]-arr2[j]
        if(abs(calc)>d):
            ans+=1
    if(len(arr2)==ans):
        count+=1
count
```
