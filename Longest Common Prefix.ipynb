```python
strs = ["dog","racecar","car"]

global_count=0
local_count=0
length=len(strs)
for i in range(len(strs[0])+1):
    prefix=strs[0][:i]

    for j in range(length):
        if(strs[j].startswith(prefix)):
            local_count+=1

        
    if(length==local_count):
        local_count=0
        global_count+=1
    else:
        break

strs[0][:global_count-1]
```
