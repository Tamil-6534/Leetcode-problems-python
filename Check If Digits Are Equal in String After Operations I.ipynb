```python

def hasSameDigits(s):
    lst=[]
    for i in range(len(s)-1):
        sum=str(int(s[i])+int(s[i+1]))
        lst.append(sum[-1])
        print(sum,lst)
    joined=''.join(lst)
    print(joined)
    if(len(joined)>2):
        hasSameDigits(joined)
    else:
        print(True) if(joined[0]==joined[1]) else print(False)

        

s = "34789"
hasSameDigits(s)
```
