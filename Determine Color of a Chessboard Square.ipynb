```python
coordinates = "c7"

black_first=['a','c','e','g']
white_first=['b','d','f','h']

coordinates0=coordinates[0]
coordinates1=coordinates[1]

if(coordinates0 in black_first) :
    if int(coordinates1)%2==1:
        print(False)
    else:
        print(True)

if(coordinates0 in white_first):
    if int(coordinates1)%2==0:
        print(False)
    else:
        print(True)
```
