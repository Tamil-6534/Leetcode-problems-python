```python
class ParkingSystem(object):

    def __init__(self, big, medium, small):
        self.big=big
        self.medium=medium
        self.small=small

    def addCar(self, carType):
        if(carType==1) :
            if self.big>0:
                self.big-=1
                return True
            else:
                return False
                
        elif(carType==2):
            if self.medium>0:
                self.medium-=1
                return True
            else:
                return False

        elif(carType==3):
            if self.small>0:
                self.small-=1
                return True
            else:
                return False

obj=ParkingSystem(1,1,0)
inputs=[1,2,3,1]
for i in inputs:
    print(obj.addCar(i))
```
