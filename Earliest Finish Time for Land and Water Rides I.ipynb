```python
landStartTime = [99]
landDuration = [59]
waterStartTime = [99, 54]
waterDuration = [85, 20]

length1 = len(landStartTime)
length2 = len(waterStartTime)
ans = []

# Land first, then water
for i in range(length1):
    land_end = landStartTime[i] + landDuration[i]
    
    for j in range(length2):
        # When land ends, when does water actually start?
        water_start = max(land_end, waterStartTime[j])
        total = water_start + waterDuration[j]
        ans.append(total)

# Water first, then land
for i in range(length2):
    water_end = waterStartTime[i] + waterDuration[i]
    
    for j in range(length1):
        # When water ends, when does land actually start?
        land_start = max(water_end, landStartTime[j])
        total = land_start + landDuration[j]
        ans.append(total)

print("Minimum:", min(ans))
print("All times:", ans)
```
