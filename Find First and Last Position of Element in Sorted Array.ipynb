```python
def binary_search(arr,target,lower,upper):
    
    if(lower<=upper):
        mid=(lower+upper)//2
        if(arr[mid]==target):
            return mid
        elif(arr[mid]<target):
            return binary_search(arr,target,mid+1,upper)
        else:
            return binary_search(arr,target,lower,mid-1)
    else:
        return -1
    
nums = [5,7,7,8,8,10]
target = 8
lst=[]
result = binary_search((nums), target, 0, len(nums) - 1)
if result != -1:
    lst.append(result)
    if(len(nums)>=2) and (nums[result]==nums[result+1]):
        lst.append(result+1)
        print(sorted(lst))
    elif(len(nums)>=2) and (nums[result]==nums[result-1]):
        if(result-1==-1):
            lst.append(len(nums)-1)
        else:
            lst.append(result-1)
        print(sorted(lst))
    else:
        lst.append(result)
        print(lst)

else:
    print(list([-1,-1]))
```
