```python
class Solution(object):
    def countPairs(self, nums, target):
        count = 0   # ✅ initialize count

        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):  # ✅ j starts from i+1
                if nums[i] + nums[j] < target:
                    count += 1
                else:
                    break   # since nums is sorted

        return count

nums = [-6, 2, 5, -2, -7, -1, 3]
target = -2

obj = Solution()
print(obj.countPairs(sorted(nums), target))
```
