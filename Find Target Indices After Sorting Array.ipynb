```python
def findtarget(nums,low,high):
    mid=(low+high)//2
    if(low<=high):
        if(nums[mid]==target):
            lst.append(mid)
            if(nums[mid+1]==target):
                lst.append(nums[mid+1])
            elif(nums[mid-1]==target):
                lst.append(mid-1)
        elif(nums[mid]<target):
            return findtarget(nums,mid+1,high)
        elif(nums[mid]>target):
            return findtarget(nums,low,mid-1)

nums = [1,2,5,2,3]
target = 2
lst=[]
nums.sort()
findtarget(nums,0,len(nums)-1)
print(sorted(lst))
```
