```python
grid = [[1,2,4],[3,3,1]]
max_lst=[]
sum=0
length=len(grid)

for i in range(length):
    grid[i].sort()
    
for j in range(len(grid[0])):
    limit=0
    while(limit<length):
        max_lst.append(grid[limit][-1])
        grid[limit].pop(-1)
        limit+=1
    sum+=max(max_lst)
    max_lst=[]

sum
```
