```python
n = 14
not_played_winners=0
lst=[]
for i in range(n):
    total_matches=n+not_played_winners
    not_played_winners=0
    if(total_matches==1):
        break
    elif(total_matches%2==0):
        n=total_matches/2
    elif(total_matches%2==1):
        n=(total_matches-1)/2
        not_played_winners=1

    lst.append(n)

print(int(sum(lst)))
```
