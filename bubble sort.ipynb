```python
def bubblesort(arr):
    for i in range(len(arr)-1):
        swap=False
        for j in range(len(arr)-1-i):
            if(arr[j]>arr[j+1]):
                swap=True
                arr[j],arr[j+1]=arr[j+1],arr[j]
        if(not swap):
            break
    return arr


arr=[34,56,3,23,6,4,675]
print(bubblesort(arr))
```
