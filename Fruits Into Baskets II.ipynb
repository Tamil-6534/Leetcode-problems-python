```python
fruits = [3,6,1]
baskets = [6,4,7]

lst=[]

for i in fruits:
    for j in baskets:
        if(j!='*') and (i<=j):
            lst.append(i)
            index=baskets.index(j)
            baskets.pop(index)
            baskets.insert(index,'*')
            break


length=len(baskets)-len(lst)
length
```
